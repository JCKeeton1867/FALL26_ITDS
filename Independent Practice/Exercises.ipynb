```python
def congruency_match(given_list):
    first = given_list[0]
    last = given_list[-1]

    if first == last:
        print(True)
    else:
        print(False)
```

```python
numbers_x = [10,20,30,40,10]
numbers_y = [75,65,35,75,30]

congruency_match(numbers_x)
congruency_match(numbers_y)
```
