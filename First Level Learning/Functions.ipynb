```python
# Function - named, reusable block of code that performs ONE task. 
# it (optionally) takes inputs (arguments) and (optionally) sends back an output (return value)
# Functions are made by: def [function name] [parameters] NEXT LINE [function description written between ''' '''] NEXT LINE [result statement]
```

```python
def convert_c_farenheit(c):
    return (9/5)*c+32
```

```python
convert_c_farenheit(18)
```

```python
# SUM OF TWO NUMBERS
def add_numbers(n1,n2):
    return n1+n2
```

```python
add_numbers(3,6)
```

```python
my_sum_func = add_numbers
```

```python
my_sum_func(4,6)
```

```python
# DIFF OF TWO NUMBERS
def diff_of_numbers(n1,n2):
    return n1-n2
```

```python
diff_of_numbers(20,6)
```

```python
scores = [100,97,68,98,85,75,60,100]
```

```python
def letter_grade(score):
    if score >= 90:
        print("A")
    elif score >=80:
        print("B")
    elif score >=70:
        print("C")
    elif score >=60:
        print("D")
    else:
        print("F")
    print(f"{score}")
```

```python
letter_grade(90)
```

```python
# This requires you to do: import math
def stats_list(alist):
    result =  max(alist), min(alist), max(alist)-min(alist), len(alist)
    return result
```

```python
len("abcd")
```

```python
stats_list(scores)
```
