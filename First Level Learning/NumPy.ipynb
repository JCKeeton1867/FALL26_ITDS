### LAMBDA EXPRESSIONS

```python
students = [{"name":"Maya", "GPA":3.9, "Major":"Math"},
            {"name":"Amara", "GPA":3.75, "Major":"Economics"},
            {"name":"Kolby", "GPA":3.99, "Major":"Math"},
            {"name":"Ava", "GPA":4.0, "Major":"Data Science"},
            {"name":"Gaby", "GPA":3.95, "Major":"AI"},
            ]
```

```python
students
```

```python
ranked = sorted(students,
                key=lambda s: (-s["GPA"], s["Major"], s["name"]))
```

```python
for s in ranked:
    print(f'{s["Major"]:<20}{s["GPA"]}{s["name"]:>10}')
```

```python
def multiply(a,b):
    return a*b
```

```python
multiply(9,30)
```

```python
ml = lambda a,b: a*b
```

```python
ml(5,8)
```

```python
def sqrn(numb):
    return numb**2
```

```python
sqrn(36)
```
