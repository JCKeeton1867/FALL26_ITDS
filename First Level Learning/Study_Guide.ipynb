```python
temps = [72, 68, 95, 81, 60, 88, 77]
total = 0
above80 = 0
largest = temps[0] # start from the first reading, not 0
for t in temps:
    total += t
    if t > 80:
        above80 += 1
    if t > largest:
        largest = t
        
print(f"Average: {total / len(temps):.2f}") # Average: 77.29
print(f"Readings above 80: {above80}") # 3
print(f"Largest reading: {largest}") # 95

print("Below 80:")
for t in temps:
    if t >= 80:
        continue
print(t) # 72, 68, 60, 77
```

```python
temps = [72, 68, 95, 81, 60, 88, 77]

def sum_of_temp(t):
    total = 0
    for t in temps:
        total = total + t
        avg = total//len(temps)
    return avg
```
