**Look at the value counts of the Gender column. Something is inconsistent. Fix it so there are only three categories.**

```python
#better way to do it
print(sal["Gender"].value_counts())
sal['Gender'] = sal['Gender'].replace({'F': 'Female', 'M': 'Male'})
sal['Gender'].value_counts()
```

```python


x = sal["Gender"]
male_count = 0
female_count = 0
nb_count = 0

for i in x:
    if i == "Male":
        male_count += 1
        continue
    elif i == "Nonbinary":
        nb_count += 1
        continue
    elif i == "M":
        male_count += 1
        continue
    elif i == "F": 
        female_count += 1
        continue
    elif i == "Female" or "F":
        female_count += 1
        continue

print(f' Male = {male_count}')
print(f' Female = {female_count}')
print(f' NonBinary = {nb_count}')

print(sal["Gender"].value_counts())
```
