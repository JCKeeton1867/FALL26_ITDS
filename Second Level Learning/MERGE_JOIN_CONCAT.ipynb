## Merge

```python
# how merge works
# combines based on the specificied column. For IDs that exist on both sides, it creates a row of combined data. For IDs that don't exist on both sides
# the row will be removed (if how is specified as "inner"). If the ID is repeated on one side more than the other, a separate row is created for each 
# instance of the ID on the most repeated side. 
```

```python
stugrad1 = students.merge(grades, on="student_id", how="inner")
```

```python
stugrad1
```

```python
# if how is defined as "outer", then rows not included on both sections will have NaN on areas not specified by one dataframe
```

```python
students.merge(grades, on="student_id", how="outer")
```

```python
# If how is defined as "left", it will put in everything on the 'left' (first mentioned) table, and put corresponding info on the right.
# If how is defined as "right", the opposite will take place. 
```

```python
students.merge(grades, on="student_id", how="left")
```

```python
students.merge(grades, on="student_id", how="right")
```

## Join

```python
s_id = students.set_index("student_id")
g_id = grades.set_index('student_id')

# no need to specify 'on' because it is always joining on the previously specified index (which was "student_id")

s_id.join(g_id)
```

```python
s_id
```

```python
g_id.join(s_id)
```

## concat

```python
fall_roster = pd.DataFrame({
    "student_id": [101, 102], "name": ["Amina", "Bilal"], "credits": [15, 12]})
spring_roster = pd.DataFrame({
    "student_id": [103, 104], "name": ["Chen", "Dalia"], "credits": [16, 14]})

print("fall_roster");   print(fall_roster)
print("\nspring_roster"); print(spring_roster)
```

```python
fall_roster
```

```python
stacked = pd.concat([fall_roster, spring_roster])
stacked
```

```python
long = pd.concat([fall_roster, spring_roster], axis=1)
long
```
