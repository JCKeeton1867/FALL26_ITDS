```python
#Q1
agev = int(float(input("Age"))) # Q1

fare = 0
cat = ""

if agev>120 or agev<0:
    print("Invalid Age")
elif 0<=agev<=4:
    fare = 0
    cat = "(Infant)"
    print(f'"Age {agev} pays ${fare} {cat}"')
elif 5<=agev<=12:
    fare = 8
    cat = "(Child)"
    print(f'"Age {agev} pays ${fare} {cat}"')
elif 13<=agev<=64:
    fare = 15
    cat = "(Adult)"
    print(f'"Age {agev} pays ${fare} {cat}"')
elif 65<=agev<=120:
    fare = 10
    cat = "(Senior)"
    print(f'"Age {agev} pays ${fare} {cat}"')
```

```python
#Q2
scores = [58, 91, 74, 66, 88, 79, 95, 61] # Q2

total = 0
above80 = 0
smallest = 10000000


for i in scores:
    total = total + i
    if i>=80:
        above80 += 1
    if i < smallest:
        smallest = i

avg = total/len(scores)

print(f' Average is {avg}')
print(f' Total is {total}')
print(f' There are {above80} scores above 80')
print(f' The smallest number is {smallest}')
```

```python
#Q3
sentence = input("please enter a sentence").lower().strip()

consonants = 0
for ch in sentence.strip():
    if ch not in "aeiou":
        consonants += 1

words = sentence.split()

word_count = len(words)

long_word = "" #finding longest word
for w in words:
    if len(w)>len(long_word):
        long_word = w
print(f' There are {consonants} consonants in this sentence')
print(f' There are {word_count} words in this sentence')
print(f' The longest word is {long_word}')
print(f' This sentence reversed is: {"_".join(words[::-1])}')
```
